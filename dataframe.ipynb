```python
import pandas as pd

df = pd.read_csv("disperkim-od_18378_jml_sampah_terangkut_ke_tempat_pembuangan_akhir_tpa__k_data.csv")
```

```python

total_sampah_2020 = 0

for index, row in df.iterrows():
    if row['tahun'] == 2020:
        total_sampah_2020 += row['jumlah_sampah']

print(f"Total sampah seluruh Kabupaten/Kota di Jawa Barat tahun 2020: {total_sampah_2020}")
```

```python
total_per_tahun = {}

for index, row in df.iterrows():
    tahun = row['tahun']
    jumlah = row['jumlah_sampah']

    if tahun not in total_per_tahun:
        total_per_tahun[tahun] = 0

    total_per_tahun[tahun] += jumlah

for tahun, total in total_per_tahun.items():
    print(f"Tahun {tahun}: Total = {total:.2f}")
```
